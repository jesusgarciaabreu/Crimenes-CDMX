La Ciudad de México (CDMX) es una de las urbes más grandes y dinámicas de América Latina, pero también enfrenta desafíos significativos en materia de seguridad pública. En este estudio, se utilizarán datos públicos oficiales para examinar la incidencia delictiva en la CDMX, identificando tendencias temporales, zonas de mayor riesgo y tipos de delito predominantes. Este enfoque cuantitativo permitirá no solo visualizar la situación actual, sino también aportar insights valiosos para autoridades, investigadores y ciudadanos interesados en la seguridad urbana.

El objetivo es transformar los datos en información accionable, contribuyendo a un debate informado sobre cómo reducir la criminalidad y construir una ciudad más segura para todos.

<h1 style="color:#00000"><strong>ANÁLISIS TEMPORAL CON SERIES DE TIEMPO.</strong></h1>

```python
from scipy.signal import find_peaks
```

```python
# Columna de fecha y hora en el formato correcto.
data_delitos_cdmx['FechaHoraInicio'] = pd.to_datetime(
    data_delitos_cdmx['FECHA DE INICIO'] + ' ' + data_delitos_cdmx['HORA DE INICIO'],
    dayfirst=True,
    format='%d/%m/%Y %H:%M'
)
```

```python
data_delitos_cdmx
```

```python
# 2) Convertir la columna a DatetimeIndex
df = data_delitos_cdmx.set_index("FechaHoraInicio")
serie_horaria = df.resample("1H").size()

serie_horaria
```

```python
# 2) FFT
y = serie_horaria.values
y_centered = y - np.mean(y)
N = len(y_centered)
yf = np.fft.fft(y_centered)
xf = np.fft.fftfreq(N, d=1)
mask = xf > 0
xf_pos = xf[mask]
yf_pos = np.abs(yf[mask])
```

```python
# 3) Detección de picos
# Elegimos picos por encima del percentil 99 para quedarnos con los más destacados
threshold = np.percentile(yf_pos, 99)
peaks, props = find_peaks(yf_pos, height=threshold)
peak_freqs = xf_pos[peaks]
peak_mags = props['peak_heights']
periods_h = 1 / peak_freqs  # periodo en horas
periods_d = periods_h / 24  # periodo en días
```

```python
# 4) Ordenar por magnitud y mostrar los top 5
order = np.argsort(peak_mags)[::-1]
top = order[:5]
results = pd.DataFrame({
    'Frecuencia (ciclos/h)': peak_freqs[top],
    'Periodo (h)': periods_h[top],
    'Periodo (días)': periods_d[top],
    'Magnitud FFT': peak_mags[top]
})
```

```python
results
```

```python
# 6) Anotar gráfica con esos picos
plt.figure(figsize=(10,5))
plt.plot(xf_pos, yf_pos, label='Espectro FFT')
plt.plot(peak_freqs[top], peak_mags[top], 'ro', label='Picos principales')
for f, m in zip(peak_freqs[top], peak_mags[top]):
    plt.text(f, m, f"{(1/f)/24:.1f}d", ha='center', va='bottom')
plt.xlabel('Frecuencia (ciclos por hora)')
plt.ylabel('Magnitud FFT')
plt.title('Espectro con picos destacados y su periodo (días)')
plt.legend()
plt.show()
```

      </ul>
    </li>
    <li>
      <strong>Ciclo semanal (~7 días ≈ 168 h)</strong>
      <ul>
        <li>Con magnitud ≈ 2 240, aparece un patrón semanal —ciertas jornadas de la semana (por ejemplo, fines de semana o lunes) registran sistemáticamente más o menos denuncias.</li>

        <li>El pico en ≈ 28 h (≈ 1,17 días) podría ser un artefacto de mezcla entre patrones diarios y semanales, o bien reflejar cambios de turno/guardia que no sean exactamente de 24 h.</li>
        <li>El pico en ≈ 8 h (≈ 0,33 días) sugiere ritmos de tres “puntas” diarias, quizá vinculadas a horarios de mayor afluencia ciudadana (mañana, tarde y noche).</li>
      </ul>
    </li>
